# Часть 1. Знакомимся с `tkinter`

## 1. Что такое `tkinter`?

Мы уже умеем писать программы, которые общаются с пользователем через `print()` и `input()`. Но большинство привычных приложений показывает **окна**, **кнопки**, **надписи** и **поля ввода**.

`tkinter` — библиотека Python, позволяющая создавать такие окна. Обычно она поставляется вместе с Python: отдельная установка через `pip` не требуется.

**Запомнить:**

- `import tkinter as tk` — подключить библиотеку и дать ей короткое имя `tk`.
- `tk.Tk()` — создать главное окно приложения.
- `mainloop()` — обрабатывать действия пользователя в обычной `.py`-программе.

> **Преподавателю:** сравните `tkinter` с набором деталей конструктора. `Tk` — основание, `Button` — кнопка, `Text` — листок для записи, `Label` — табличка с надписью.

## 7. Добавляем первую кнопку

Кнопка создаётся командой `tk.Button(...)`.

- `main_window` — окно, в которое мы кладём кнопку.
- `text="Сказать привет"` — надпись на кнопке.
- `command=...` — действие при нажатии.

## 8. Учимся вводить текст — `tk.Text`

Пока кнопка всегда произносит одну и ту же фразу. А что, если пользователь хочет написать **своё** сообщение?

Для этого нужен `Text` — поле для ввода текста.

- `width=15` — примерная ширина поля в символах.
- `height=1` — одна строка по высоте.
- `.get("1.0", tk.END)` — прочитать текст от первой строки, первого символа до конца.
- `.strip()` — убрать лишние пробелы и символ перевода строки с краёв.

Чтобы поле появилось **над кнопкой**, сначала удалим старую кнопку и создадим элементы в нужном порядке: `Text`, затем `Button`. Так работает `pack()`.

````python
hello_button.destroy()

tk.Label(main_window, text="Напиши своё сообщение:").pack(pady=(15, 5))
text_box = tk.Text(main_window, width=25, height=1)
text_box.pack(pady=5)

message_button = tk.Button(
    main_window,
    text="Показать сообщение",
    command=lambda: print(text_box.get("1.0", tk.END).strip())
)
message_button.pack(pady=10)

print("Напечатайте сообщение и нажмите кнопку.")
````

### Проверяем, как работает `get()`

В `Text` позиция `1.0` означает «первая строка, символ с номером 0». `tk.END` означает «до конца текста». После чтения `Text` обычно добавляет перевод строки, поэтому для красивого вывода мы используем `.strip()`.

Введите несколько слов в поле и запускайте следующую ячейку сколько угодно раз.

````python
entered_text = text_box.get("1.0", tk.END).strip()
print("Пользователь написал:", entered_text)
````

# Часть 2. Создаём окно авторизации

## 10. Что такое авторизация?

Мы делаем учебное окно входа: пользователь вводит **логин** и **пароль**, а программа проверяет, есть ли такой ученик в списке, и совпадает ли пароль.

**Пример:**

- Логин: `admin`
- Пароль: `admin`
- Результат: `Привет — Михаил Ануфрев`

Если логина нет или пароль не подходит, выводим **«Доступ запрещен»**.

**Правило безопасности:** это только учебный тренажёр! В примере пароли специально придуманы и хранятся в открытом виде ради объяснения алгоритма. Так **нельзя** хранить пароли настоящего сайта. Не вводите здесь свои настоящие пароли.

> **Вопрос:** зачем приложению вообще проверять и логин, и пароль? Почему одного логина недостаточно?

## 11. Из каких частей состоит проект? Принцип MVC

В методичке используется разделение кода на части **MVC (Model–View–Controller)**. Думайте об этом как о работе команды:

| Часть | На понятном языке | Наш файл |
|---|---|---|
| **Model** | Хранит информацию об одном пользователе | `User.py` |
| **Repository** | Ищет нужного пользователя среди всех | `user_repository.py` |
| **Controller** | Решает, разрешить вход или нет | `controller.py` |
| **View** | Показывает окно, поля и кнопку | `view.py` |
| **Config** | Хранит настройки окна | `config.py` |

**Аналогия:** в школьной библиотеке есть карточки читателей (Model), картотека (Repository), библиотекарь, который проверяет карточку (Controller), и стойка, за которой всё происходит (View).

> **Преподавателю:** слово MVC вначале можно дать без запоминания аббревиатуры. Главная идея — не писать всю большую программу в одном файле.

## 12. Класс пользователя — `User.py`

Нам нужен способ описать **одного** человека. Для этого создадим класс `User`.

Что здесь новое:

- `class User:` — описание будущих пользователей.
- `__init__` — вызывается, когда создаётся новый пользователь.
- `self` — означает «этот конкретный пользователь».
- `_login`, `_password`, `_fio` — сохранённые данные. Нижнее подчёркивание напоминает: не меняйте их снаружи без необходимости.
- `check_password(password)` — сравнивает пароль.
- `get_name()` — возвращает имя.
- `get_login()` — возвращает логин.

**Сравнение:** класс — пустая карточка с графами, а `User("admin", "admin", "Михаил Ануфрев")` — заполненная карточка одного человека.

````python
user_code = '''class User:
    def __init__(self, login, password, fio):
        self._login = login
        self._password = password
        self._fio = fio

    def check_password(self, password):
        return password == self._password

    def get_name(self):
        return self._fio

    def get_login(self):
        return self._login
'''

(PROJECT_DIR / "User.py").write_text(user_code, encoding="utf-8")
print("Файл User.py создан!")
````

### Проверяем модель без окна

Теперь код живёт в отдельном файле, но его класс можно импортировать в ноутбук. Создадим одного тестового пользователя и вызовем его методы.

**Обратите внимание:** `check_password` возвращает **`True`** или **`False`**, а не пишет сообщение самостоятельно.

````python
from User import User

student = User("cat", "123", "Кот Программист")
print("Логин:", student.get_login())
print("Имя:", student.get_name())
print("Пароль 123 верный?", student.check_password("123"))
print("Пароль 321 верный?", student.check_password("321"))
````

## 13. Где хранятся пользователи — `user_repository.py`

Хранить одного человека мы умеем. Но приложению нужен список **нескольких** людей, а также поиск по логину.

В методичке используются **пять учебных пользователей**. Оставим их, чтобы полностью проверить исходное задание.

- `users` — список объектов `User`.
- `users_repos` — словарь: **ключ** это логин, **значение** это объект пользователя.
- `get_user_by_login(login)` — найти пользователя или вернуть `None`, если его нет.

**Аналогия словаря:** у каждого шкафчика есть номер (логин), по которому можно сразу найти нужную карточку.

````python
repository_code = '''from User import User

# Пять выдуманных учебных учётных записей из методички.
users = [
    User("admin", "admin", "Михаил Ануфрев"),
    User("winter", "is_near", "Джон Сноу"),
    User("jane", "password", "Джейн Смит"),
    User("bob", "secret123", "Боб Джонс"),
    User("susan", "p@ssw0rd", "Сьюзан Вонг"),
]

# Собираем словарь: логин -> объект User.
users_repos = {user.get_login(): user for user in users}

# После создания словаря список уже не нужен.
users.clear()


def get_user_by_login(login):
    return users_repos.get(login)
'''

(PROJECT_DIR / "user_repository.py").write_text(repository_code, encoding="utf-8")
print("Файл user_repository.py создан!")
````

### Словарь — маленький шаг в сторону

Строка `users_repos = {user.get_login(): user for user in users}` может показаться сложной. Она означает «пройти по всем пользователям и записать каждого в словарь по его логину».

То же самое можно записать более длинно:

```python
users_repos = {}
for user in users:
    users_repos[user.get_login()] = user
```

А `users.clear()` удаляет элементы из **исходного списка**, но уже созданный **словарь** сохраняет сами объекты.

> **Преподавателю:** детям 10–12 лет достаточно понять идею словаря «логин → пользователь»; не требуйте запоминать генератор словаря с первого раза.

## 14. Кто проверяет логин и пароль — `controller.py`

Теперь сделаем функцию `auth_user(login, password)`. Она отвечает на один вопрос: **можно ли впустить человека?**

Алгоритм очень простой:

- Убрать случайные пробелы вокруг логина: `.strip()`.
- Найти пользователя по логину.
- Сначала предположить, что вход **запрещён**.
- Если пользователь найден, проверить пароль через `check_password()`.
- При успехе показать `Привет — ...`, иначе `Доступ запрещен`.

В исходной методичке используется `pass` — команда «ничего не делать». Мы сохраняем её, чтобы вы увидели знакомую конструкцию. В обычном проекте эту проверку можно записать короче.

**Небольшое улучшение:** помимо `print(message)` функция ещё и **возвращает** сообщение. Так мы сможем показать ответ прямо в окне.

````python
controller_code = '''import user_repository


def auth_user(login, password):
    user = user_repository.get_user_by_login(login.strip())
    message = "Доступ запрещен"

    if user is None:
        pass  # Пользователь не найден — сообщение не меняем.
    else:
        if user.check_password(password.strip()):
            message = f"Привет — {user.get_name()}"

    print(message)
    return message
'''

(PROJECT_DIR / "controller.py").write_text(controller_code, encoding="utf-8")
print("Файл controller.py создан!")
````

- `admin` + `wrong` → отказ.

- `  jane  ` + `password` → приветствие, потому что `.strip()` удаляет пробелы вокруг логина.

## 20. Создаём полноценный файл `view.py`

До этого мы собирали форму по отдельным ячейкам. Теперь **сохраним тот же интерфейс в один файл**.

В нашей версии есть функция `create_window()`, которая собирает окно. Так ноутбук может при необходимости вызывать её, а отдельный запуск `python view.py` всё равно работает привычным способом через `mainloop()`.

Отличия от исходной методички небольшие и объяснимые:

- Вместо `from config import *` используем явные имена.
- Иконка необязательна, чтобы программа запускалась на разных компьютерах.
- Внутри окна выводится сообщение о результате входа (в методичке оно печатается в консоль).
- Сам алгоритм проверки и учебная база из методички сохранены.

````python
view_code = '''import tkinter as tk
from pathlib import Path
from config import TITLE, WIDTH, HEIGHT, ICON
import controller


def create_window():
    # 1. Создаём окно и задаём размеры.
    main_window = tk.Tk()
    main_window.title("Авторизация — " + TITLE)
    main_window.geometry(f"{WIDTH}x{HEIGHT}")

    # 2. Устанавливаем иконку, если система её поддерживает.
    icon_path = Path(__file__).with_name(ICON)
    if icon_path.exists():
        try:
            main_window.iconbitmap(str(icon_path))
        except tk.TclError:
            pass

    # 3. Поле логина.
    tk.Label(main_window, text="Логин", font=("Arial", 12)).pack(pady=(20, 5))
    login_text = tk.Text(main_window, width=20, height=1, font=("Arial", 12))
    login_text.pack(pady=5)

    # 4. Поле пароля. В методичке также используется Text.
    tk.Label(main_window, text="Пароль", font=("Arial", 12)).pack(pady=(12, 5))
    password_text = tk.Text(main_window, width=20, height=1, font=("Arial", 12))
    password_text.pack(pady=5)

    # 5. Место, где будем показывать результат.
    result_label = tk.Label(main_window, text="", font=("Arial", 12))

    def on_auth_click():
        login = login_text.get("1.0", tk.END)
        password = password_text.get("1.0", tk.END)
        message = controller.auth_user(login, password)
        result_label.config(text=message)

    # 6. Кнопка запускает проверку.
    auth_button = tk.Button(
        main_window, text="Войти", font=("Arial", 12), command=on_auth_click
    )
    auth_button.pack(pady=15)
    result_label.pack(pady=5)

    return main_window


if __name__ == "__main__":
    create_window().mainloop()
'''

(PROJECT_DIR / "view.py").write_text(view_code, encoding="utf-8")
print("Файл view.py создан!")
````

| `No module named tkinter` | Проверить установку Python с поддержкой Tk. Библиотека не ставится обычной командой `pip install tkinter`. |
| `TclError: no display name` | Запускать локальный Jupyter на компьютере с рабочим столом, а не на сервере без графического экрана. |
| `invalid command name` / окно исчезло | Возможно, окно закрыли. Снова выполните ячейку, которая его создаёт. |
| Иконка не меняется | Проверьте имя `main_window.ico`, папку проекта и поддержку иконок системой. Это не мешает остальному уроку. |
| Кнопка ничего не делает | Проверьте `command=...` без лишних круглых скобок, а также `.pack()`. |
| Пользователь не находится | Проверьте логин, регистр букв и запись пользователя в `user_repository.py`. |
| Изменения в `.py` не видны | Закройте приложение и запустите `view.py` снова. |

> **Самая частая ошибка:** `command=on_auth_click()` сразу вызывает функцию при создании кнопки. Нужно `command=on_auth_click` — без скобок, чтобы функция запускалась **по нажатию**.
